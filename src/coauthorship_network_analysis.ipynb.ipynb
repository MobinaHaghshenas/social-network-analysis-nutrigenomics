```python
import networkx as nx
import pandas as pd
df = pd.read_csv('csv-nutrigenom-pubmed.csv')
```

```python
df.isna().sum()
```

```python
df = df.dropna(subset=['Authors'])
```

```python
df.isna().sum()
```

```python
df.shape
```

```python
G = nx.Graph()

for index, row in df.iterrows():
    for author in row['Authors'].split(','):
        author = author.strip()  
        G.add_node(author, bipartite=0)  # authors are in set 0

pmids = df['PMID'].unique()
G.add_nodes_from(pmids, bipartite=1)  # PMIDs are in set 1

for index, row in df.iterrows():
    for author in row['Authors'].split(','):
        author = author.strip()  
        G.add_edge(author, row['PMID'])

print(G)

print("Author nodes:")
for node in G.nodes(data=True):
    if node[1]['bipartite'] == 0:
        print(node[0])
```

```python
print("Author nodes:")
authors = [node for node in G.nodes(data=True) if node[1]['bipartite'] == 0]
for author in authors:
    print(author[0])

print("\nPMID nodes:")
pmids = [node for node in G.nodes(data=True) if node[1]['bipartite'] == 1]
for pmid in pmids:
    print(pmid[0])
```

```python
print("Author nodes:")
for node in G.nodes(data=True):
    if node[1]['bipartite'] == 0:
        print(node[0])
```

```python
coauthor_graph = nx.Graph()

for edge in G.edges(data=False):
    author1 = edge[0]
    pmid = edge[1]
    for other_edge in G.edges(data=False):
        if other_edge[1] == pmid:  
            author2 = other_edge[0]
            if author1 != author2: 
                coauthor_graph.add_edge(author1, author2)

print(coauthor_graph)
```

```python
betweenness_centrality = nx.betweenness_centrality(coauthor_graph)

print(betweenness_centrality)
```

```python
total_betweenness_centrality = sum(betweenness_centrality.values())
average_betweenness_centrality = total_betweenness_centrality / len(coauthor_graph.nodes())
print(average_betweenness_centrality)
```

```python
import numpy as np

pr = nx.pagerank(coauthor_graph, alpha=0.85, personalization=None, max_iter=1000, tol=1e-6)
print(np.mean(list(pr.values())))
```

```python
degree_centrality = nx.degree_centrality(coauthor_graph)

sorted_authors = sorted(degree_centrality.items(), key=lambda x: x[1], reverse=True)
print("Top 10 authors by degree centrality:")
for author, centrality in sorted_authors[:10]:
    print(f"{author}: {centrality}")
```

```python
betweenness_centrality = nx.betweenness_centrality(coauthor_graph)

sorted_authors = sorted(betweenness_centrality.items(), key=lambda x: x[1], reverse=True)
print("Top 10 authors by betweenness centrality:")
for author, centrality in sorted_authors[:10]:
    print(f"{author}: {centrality}")
```

```python
import numpy as np
degree_centrality = nx.degree_centrality(coauthor_graph)
print("degree centrality :", degree_centrality)
median_degree_centrality = np.median(list(degree_centrality.values()))
print(f"Median degree centrality of the network: {median_degree_centrality}")
```

```python
import networkx as nx
import numpy as np

degrees = [d[1] for d in nx.degree(coauthor_graph, weight='weight')]
average_degree = np.mean(degrees)
print(f"Average weighted degree of the network: {average_degree}")
```
